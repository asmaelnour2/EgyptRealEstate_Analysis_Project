```python
import pandas as pd
import re

df = pd.read_csv('egypt_real_estate_listings.csv')
```

```python
# --- (Silver Layer) ---
# Price Transformation
df['price'] = pd.to_numeric(df['price'].str.replace(',', '', regex=False), errors='coerce')
```

```python
#  Size Transformation
df['size_sqm'] = df['size'].str.extract(r'(\d+,?\d*)\s*sqm')[0].str.replace(',', '', regex=False).astype(float)
```

```python
#Create "has_maid" flag (Yes/No) 
df['has_maid'] = df['bedrooms'].str.contains('Maid', case=False, na=False).map({True: 'Yes', False: 'No'})
```

```python
#  Create "is_studio" flag (Yes/No)
df['is_studio'] = df['bedrooms'].str.contains('Studio', case=False, na=False).map({True: 'Yes', False: 'No'})
```

```python
#  Bedrooms Cleaning: 
df['bedrooms_count'] = df['bedrooms'].str.extract(r'(\d+)').astype(float)
df.loc[df['is_studio'] == 'Yes', 'bedrooms_count'] = 1.0
```

```python
#  Bathrooms Cleaning convert to float
df['bathrooms_count'] = df['bathrooms'].str.extract(r'(\d+)').astype(float)
```

```python
# Extract Governorate and City from the location string
df['governorate'] = df['location'].str.split(',').str[-1].str.strip()
df['city'] = df['location'].str.split(',').str[-2].str.strip()
```

```python
#  Convert availability string to actual DateTime objects
df['available_from'] = pd.to_datetime(df['available_from'], errors='coerce')
```

```python
# Drop old/redundant columns to optimize memory
cols_to_drop = ['description', 'down_payment', 'size', 'location', 'bedrooms', 'bathrooms']
df.drop(columns=cols_to_drop, inplace=True, errors='ignore')
```

```python
#  Remove records with missing critical values (Price/Size)
df = df.dropna(subset=['price', 'size_sqm'])
```

```python
# --- Final Output  ---
print("ETL Pipeline Completed Successfully!")
print("\n--- Final Schema (Data Types) ---")
print(df.dtypes)
```
