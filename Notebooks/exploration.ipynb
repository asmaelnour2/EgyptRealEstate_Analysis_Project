```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('egypt_real_estate_listings.csv')
```

```python
# --- 1. General Overview ---
print("--- 1. Shape of the Dataset ---")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")

print("\n--- 2. Data Types and Non-Null Counts ---")
# Check how many missing values in each column and their data types
print(df.info())
```

```python
# --- 2. Statistical Profiling ---
print("\n--- 3. Summary Statistics (All Columns) ---")
print(df.describe(include='all'))
```

```python
# --- 3. Missing Values ---
print("\n--- 4. Missing Values Analysis ---")
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_report = pd.concat([missing_count, missing_percentage], axis=1, keys=['Missing Count', '%'])
print(missing_report.sort_values(by='%', ascending=False))
```

```python
# --- 4. Exploring Categorical Columns (Data Quality Check) ---
print("\n--- 5. Unique Values in Categorical Columns ---")
categorical_cols = ['type', 'payment_method', 'bedrooms', 'bathrooms']
for col in categorical_cols:
    print(f"Unique values in '{col}': {df[col].unique()[:10]}...") # Showing first 10 for brevity
```

```python
# A. Inspecting 'price' (Target Column)
print("\n--- 6. Price Column Sample (Initial State) ---")
print(df['price'].head(10).to_list())
```

```python
# B. Inspecting 'location'
print("\n--- 7. Location Structure Analysis ---")
comma_counts = df['location'].str.count(',').value_counts()
print("Number of commas in location field distribution:")
print(comma_counts)
```

```python
# C. Inspecting 'size' (Unit Inconsistency)
print("\n--- 8. Size Column Sample ---")
print(df['size'].head(5).to_list())
```
